# Prueba Kolmogorov-Smirnov (K-S)

La prueba de Kolmogorov-Smirnov es una prueba no paramétrica que permite
verificar si una muestra proviene de una distribución específica.

En este caso queremos comprobar si nuestros 100 datos siguen una
distribución Uniforme U(0,1).

Hipótesis:

H₀: Los datos siguen una distribución uniforme U(0,1).
H₁: Los datos NO siguen una distribución uniforme.

Procedimiento:

1. Ordenar los datos de menor a mayor.
2. Calcular:
   
   $D^+ = \max \left\{ \frac{i}{n} - x_{(i)} \right\}$.

   $D^- = \max \left\{ x_{(i)} - \frac{i-1}{n} \right\}$.

3. Obtener $D = max(D⁺, D⁻)$
4. Comparar D con el valor crítico Dα,n.

## Implementación manual de la prueba K-S

Primero ordenamos los datos.
Después calculamos $\frac{i}{n}$ y $\frac{i-1}{n}$.
Luego obtenemos los vectores D⁺ y D⁻.
Finalmente determinamos el estadístico D.

```python
def prueba_ks(data, alfa=0.05):

    # Ordenamos los datos
    data_sorted = np.sort(data)
    n = len(data)

    # Creamos el vector i
    i = np.arange(1, n + 1)

    # Calculamos i/n y (i-1)/n
    i_n = i / n
    i_1_n = (i - 1) / n

    # Calculamos D+ y D-
    d_mas_vec = i_n - data_sorted
    d_menos_vec = data_sorted - i_1_n

    # Tomamos los máximos
    d_mas = np.max(d_mas_vec)
    d_menos = np.max(d_menos_vec)

    # Estadístico final
    d_calc = max(d_mas, d_menos)

    # Valor crítico (tabla aproximada usando scipy)
    d_critico = st.kstwo.ppf(1 - alfa, n)

    print("Resultados Finales:")
    print(f"D+ máximo: {d_mas:.5f}")
    print(f"D- máximo: {d_menos:.5f}")
    print(f"D calculado: {d_calc:.5f}")
    print(f"D crítico: {d_critico:.5f}")

    if d_calc < d_critico:
        print("No se rechaza H0")
    else:
        print("Se rechaza H0")

    return d_calc, d_critico
```

```python
prueba_ks(data, 0.05)
```
