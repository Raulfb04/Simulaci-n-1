Ya que Python lee los datos por filas (es decir, de izquierda a derecha), es necesario reorganizarlos para que se lean por columnas.
Para lograrlo, primero convertimos la tabla de datos en una matriz; posteriormente transponemos dicha matriz y finalmente la volvemos a convertir en un vector.
De esta manera, los datos se procesan de arriba hacia abajo, que es el orden que necesitamos para este ejemplo.

```python
import numpy as np

data_array = np.array(data)

matriz = data_array.reshape(20, 5)

matriz_T = matriz.T

data_columna = matriz_T.flatten()

data_columna
```

Contamos el número de rachas observadas

```python
def contar_rachas(data_columna):

    signos = []

    for i in range(1, len(data_columna)):
        if data_columna[i] > data_columna[i-1]:
            signos.append("+")
        elif data_columna[i] < data_columna[i-1]:
            signos.append("-")

    rachas = 1

    for i in range(1, len(signos)):
        if signos[i] != signos[i-1]:
            rachas += 1

    print("Secuencia de signos:")
    print(signos)
    print("\nNúmero total de rachas:", rachas)

    return rachas, signos
```

```python
rachas, signos = contar_rachas(data_columna)
```

Ahora vamos con la parte estadistica, en donde calcularemos el valor esperado, la varianza y el estadistico Z:

$\mu_{c_o} = \frac{2 n - 1}{3}$

$\sigma_{c_o}^2 = \frac{16n - 29}{90}$

$Z_0 = \frac{c_0 - \mu_{c_o}}{\sigma_{c_o}}$



```python
n = len(data_columna)
c_0 = rachas

# Media esperada
mu_c0 = (2*n - 1) / 3

# Varianza
var_c0 = (16*n - 29) / 90

# Desviación estándar
sigma_c0 = np.sqrt(var_c0)

# Estadístico Z
Z_0 = (c_0 - mu_c0) / sigma_c0

print("Resultados estadísticos:")
print(f"Media esperada: {mu_c0:.4f}")
print(f"Varianza: {var_c0:.4f}")
print(f"Desviación estándar: {sigma_c0:.4f}")
print(f"Z calculado: {Z_0:.4f}")
```

Para completar la prueba, calculamos el valor crítico
$z_{\alpha/2}$ de la distribución normal estándar.

La regla de decisión es:

- Si $|Z₀| ≤ z_{\alpha/2}$ → No se rechaza $H_0$.
- Si $|Z₀| > z_{\alpha/2}$ → Se rechaza $H_0$.

```python
from scipy.stats import norm

alfa = 0.05

# Valor crítico bilateral
z_critico = norm.ppf(1 - alfa/2)

print(f"Z crítico (±): {z_critico:.4f}")

# Comparación
if abs(Z_0) <= z_critico:
    print("No se rechaza H0: La secuencia es aleatoria.")
else:
    print("Se rechaza H0: La secuencia NO es aleatoria.")
```
